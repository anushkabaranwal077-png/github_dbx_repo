```python
%sql
UPDATE delta.`/Volumes/deltalake/default/deltavol/dmlSnk/`
SET salary = 1000
where id = 8;
    
Select * from delta.`/Volumes/deltalake/default/deltavol/dmlSnk/`
ORDER BY ID

'''
data1 = [1, 2, 3, 4]
data2 = [8, 5, 6, 7]
both saved inside a single table
final table has = [1, 2, 3, 4, 8, 5, 6, 7]
Update => we get deletion vector
And additional json file is also getting added.
Process:
1st : data1 json (read)
2nd : data2 json (read)
remove 2 : only records which were changed later (previous val of salary)
add 3rd : Update
Add 2 using deletion vector. (Here everything apart from the changed record will be added)
At last all the json files will be combined to 1 by Databricks automatically which is the reason why we see additional json file.
'''

```
