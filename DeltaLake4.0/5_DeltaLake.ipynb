# Delta Log Level Schema Changes

```python
data = [(1, 100, 'xyz', 1), (9, 200, 'bb', 1), (10, 300, 'cc', 1)]
df = spark.createDataFrame(data, ['cust_id', 'income', 'name', 'tip'])
display(df)
```

```python
df.write.format("delta")\
    .mode("append")\
    .save("/Volumes/deltalake/default/deltavol/schemalevel/")
```

```python
%sql
ALTER TABLE delta.`/Volumes/deltalake/default/deltavol/schemalevel/`
RENAME COLUMN name to customer_name
```

```python
%sql
ALTER TABLE delta.`/Volumes/deltalake/default/deltavol/schemalevel/`
SET TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name'
)
-- You need to enable column mapping on your Delta table before you can rename a column.
```
