### **Schema Inforcement and Schema Evolution**

```python
# Sample Data
data = [(1, 100, 'aa', 1), (2, 200, 'bb', 1), (3, 300, 'cc', 1), (4, 400, 'dd', 1)]

# Create a Spark DataFrame
df = spark.createDataFrame(data, ['id', 'salary', 'name', 'sus_col'])

display(df)
```

```python
# Since there is a schema mismatch as new columns are getting added above inside the previously created table

df.write.format("delta")\
    .mode("append")\
    .save("/Volumes/deltalake/default/deltavol/demosink/")

# Below we can se how to evolve schema
# This step is going to fail since there is a schema mismatch
```

## **SCHEMA EVOLUTION**

```python

## mergeSchema is used to evolve schema inside Delta Lake tables.

df.write.format("delta")\
    .mode("append")\
    .option("mergeSchema", "true")\
    .save("/Volumes/deltalake/default/deltavol/demosink/")
```

### **Read data from DELTA TABLE**

```python
%sql
-- Now Delta Lake will behave as a normal table.
Select * from delta.`/Volumes/deltalake/default/deltavol/demosink/`
```

# **SCHEMA OVERWRITE**

```python
# Now I want to overwrite this data. Along with it's schema.
# So I have to update the using options to overwrite it's schema
# Use mode to overwrite it's data.

df.write.format("delta")\
    .mode("overwrite")\
    .option("overwriteSchema", True)\
    .save("/Volumes/deltalake/default/deltavol/demosink/")
```

```python
%sql
-- Now Delta Lake will behave as a normal table.
Select * from delta.`/Volumes/deltalake/default/deltavol/demosink/`
```
