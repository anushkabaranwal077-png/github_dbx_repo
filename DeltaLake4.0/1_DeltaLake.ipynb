## **Create Delta Table**

### **Using SQL API** (There are 2 ways)

```python
%sql
CREATE TABLE deltalake.default.first_table
(
  id INT,
  Salary Double
)
```

### **Delta API** (2nd Way to create Delta tables)

```python
from delta.tables import DeltaTable, IdentityGenerator
from pyspark.sql.types import LongType
```

```python
DeltaTable.createIfNotExists(spark) \
  .tableName("deltalake.default.firstDeltaAPI") \
  .addColumn("id", "INT") \
  .addColumn("firstName", "STRING") \
  .addColumn("lastName", "STRING", comment = "surname") \
  .execute()
```

```python
%sql
CREATE TABLE deltalake.default.second_table (
    id INT NOT NULL,
    firstName STRING,
    middleName STRING NOT NULL,
    lastName STRING,
    gender STRING,
    birthDate TIMESTAMP,
    ssn STRING,
    salary INT
) USING DELTA;

ALTER TABLE deltalake.default.second_table CHANGE COLUMN middleName DROP NOT NULL;
```

# **Generate Columns**

### **Identity Columns** (Only present inside Delta API)

```python
%sql
Drop table deltalake.default.firstDeltaAPI
```

```python
from delta.tables import DeltaTable, IdentityGenerator
from pyspark.sql.types import LongType,IntegerType,StringType

DeltaTable.create(spark) \
  .tableName("deltalake.default.firstDeltaAPI") \
  .addColumn("id_col1", dataType=LongType(), generatedAlwaysAs=IdentityGenerator()) \
  .addColumn("salary", dataType= IntegerType()) \
  .addColumn("name", dataType= StringType()) \
  .execute()
```

### **Computed Columns**

```python
DeltaTable.create(spark) \
  .tableName("deltalake.default.firstDeltaAPI2") \
  .addColumn("salaryAfterTax", dataType=LongType(), generatedAlwaysAs="CAST((salary * 0.7) AS BIGINT)") \
  .addColumn("salary", dataType= IntegerType()) \
  .addColumn("name", dataType= StringType()) \
  .execute() 
```

```python
%sql
INSERT INTO deltalake.default.firstDeltaAPI2(salary, name)
VALUES 
(100, 'AA'),
(200, 'BB'),
(300, 'CC')
```
