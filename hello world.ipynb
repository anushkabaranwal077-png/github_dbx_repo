```python
def add_numbers(a, b):
    return a + b

def find_Even(a, b):
    if a % 2 == 0 and b % 2 == 0:
        return True
    else:
        return False

result = add_numbers(3, 5)
result2 = find_Even(4, 6)
result3 = find_Even(3, 6)
print(result, result2, result3)
```
